## 1. Introduction to Regex

A **Regular Expression (Regex)** is a pattern used to find or work with specific text.

Examples:
- Find a phone number in a paragraph.
- Extract all email addresses from a file.
- Check whether a string contains only digits.
- Replace unwanted characters in text.

Python provides the built-in `re` library for regex operations.

```python
import re

text = "My contact number is 9876543210"
result = re.search(r"\d{10}", text)

print(result.group())
```

## 2. Core Elements of Regex

| Pattern | Meaning | Example |
|---|---|---|
| `.` | Any character except newline | `c.t` |
| `^` | Starts with | `^Hello` |
| `$` | Ends with | `world$` |
| `*` | Zero or more repetitions | `ab*` |
| `+` | One or more repetitions | `ab+` |
| `?` | Zero or one repetition | `colou?r` |
| `{n}` | Exactly n repetitions | `\d{4}` |
| `{n,m}` | Between n and m repetitions | `\d{2,4}` |
| `[]` | Any one character in the set | `[aeiou]` |
| `[^]` | Any one character not in the set | `[^0-9]` |
| `\d` | Digit (0–9) | `\d+` |
| `\D` | Non-digit | `\D+` |
| `\w` | Letter, digit, or underscore | `\w+` |
| `\W` | Non-word character | `\W` |
| `\s` | Whitespace | `\s+` |
| `\S` | Non-whitespace | `\S+` |
| `\b` | Word boundary | `\bcat\b` |
| `|` | OR condition | `cat|dog` |
| `()` | Grouping | `(ab)+` |

Use a raw string such as `r"\d+"` for regex patterns. It helps Python interpret backslashes correctly.

## 3. Basic Regex Functions

Common functions from the `re` library:

- `re.search()` – finds the first match anywhere in the text.
- `re.match()` – checks for a match at the beginning.
- `re.fullmatch()` – checks whether the entire string matches.
- `re.findall()` – returns all matching values.
- `re.finditer()` – returns match objects for all matches.
- `re.sub()` – replaces matching text.
- `re.split()` – splits text using a pattern.

```python
text = "Python is easy. Python is powerful."

# Find the first occurrence
print(re.search(r"Python", text).group())

# Find all occurrences
print(re.findall(r"Python", text))

# Replace a word
print(re.sub(r"Python", "NLP", text))

# Split using spaces
print(re.split(r"\s+", text))
```

## 4. Character Sets and Quantifiers

A **character set** matches one character from a given group. A **quantifier** tells regex how many times a pattern can repeat.

```python
text = "My marks are 85, 90, 76 and 100."

# Find all numbers
print(re.findall(r"\d+", text))

# Find two-digit numbers only
print(re.findall(r"\b\d{2}\b", text))

# Find words starting with a capital letter
print(re.findall(r"\b[A-Z][a-z]*\b", "Ravi and Neha study Python."))
```

## 5. Groups and OR Operator

Parentheses `()` create groups. The pipe `|` means OR.

```python
text = "I use Python and SQL."

# Find either Python or SQL
print(re.findall(r"Python|SQL", text))

# Extract day, month and year as separate groups
date = "Date: 25-12-2026"
result = re.search(r"(\d{2})-(\d{2})-(\d{4})", date)

print("Day:", result.group(1))
print("Month:", result.group(2))
print("Year:", result.group(3))
```

## 6. Practical Example: Extract Email Addresses

The pattern below is a beginner-friendly example for extracting common email formats. It is useful for practice, but it is not a complete email-standard validator.

```python
text = """
Contact us:
support@gmail.com
harshit@yahoo.com
invalid-email
team@datamites.com
"""

emails = re.findall(r"[\w.-]+@[\w.-]+\.\w+", text)
print(emails)
```

## 7. Practical Example: Extract Indian Mobile Numbers

This example looks for 10-digit mobile numbers beginning with 6, 7, 8, or 9.

```python
text = "Call 9876543210 or 8123456789. Code 1234567890 is not a mobile number."

mobile_numbers = re.findall(r"\b[6-9]\d{9}\b", text)
print(mobile_numbers)
```

## 8. Practical Example: Clean Text

Regex can remove punctuation, extra spaces, and digits when preparing text for NLP.

```python
text = "Hello!!!   Welcome to NLP 2026. Let's learn."

# Keep letters and spaces
clean_text = re.sub(r"[^A-Za-z\s]", "", text)

# Replace multiple spaces with one space
clean_text = re.sub(r"\s+", " ", clean_text).strip()

print(clean_text)
```
